#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# Load the tidyverse package
library(tidyverse)

# Select the variables needed
tidy_march_madness <- march_madness |>
    select(year, school, seed, reg_wins_pct, tourney_finish) |>

    # Remove rows with missing values in the relevant columns
    filter(!is.na(seed),
           !is.na(reg_wins_pct),
           !is.na(tourney_finish)) |>

    # Create a categorical variable for eaching the top eight 
    mutate(top_eight = as_factor(tourney_finish %in% c("top_8_loss", "top_4_loss","top_2_loss","champ"))) |>
    # Remove the original tournament finish column
    select(year, school, seed, reg_wins_pct, top_eight)

# Display the first six rows
head(tidy_march_madness)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.

```r
# Calculate summary statistics for all numeric columns
summary_stats <- tidy_march_madness |>
    select(year, seed, reg_wins_pct) |>
    pivot_longer(
        cols = everything(),
        names_to = "variable",
        values_to = "value") |>
    group_by(variable)  |>
    summarize(
        mean = mean(value, na.rm = TRUE),
        min = min(value, na.rm = TRUE),
        max = max(value, na.rm = TRUE)) |>
    ungroup()

#Display
summary_stats
```

```r
win_pct_plot <- tidy_march_madness |>
    ggplot(aes(x = reg_wins_pct)) +
    geom_histogram(bins = 20, fill = "steelblue", colour = "white") +
    xlab("Regular-season win percentage") +
    ylab("Number of team appearances") +
    ggtitle("Distribution of Regular-Season Win Percentage") +
    theme(text = element_text(size = 14))

win_pct_plot
                   
```

```r
top_eight_summary <- tidy_march_madness |>
    group_by(top_eight) |>
    summarize(count = n()) |>
    ungroup() |>
    mutate(proportion = count / sum(count))

top_eight_plot <- top_eight_summary |>
    ggplot(aes(x = top_eight, y = proportion)) +
    geom_bar(stat = "identity", fill = "steelblue") +
    xlab("Reached the top eight (FALSE = No, TRUE = Yes)")+
    ylab("Proportion of team appearances") +
    ggtitle("Proportion of Teams Reaching the Top Eight") +
    theme(text = element_text(size = 14))

top_eight_plot
```

Visualization 1: This plot shows the proportion of teams reaching the top eight at each seed. It helps assess whether seed is useful for predicting tournament success. Differences in these proportions would suggest an association between seed and reaching the top eight, although they would not establish causation.

```r
# Calculate the proportion reaching the top eight for each seed
seed_summary <- tidy_march_madness |>
    group_by(seed) |>
    summarize(
        proportion = mean(top_eight == "TRUE")) |>
    ungroup()
seed_plot <- seed_summary |>
    ggplot(aes(x = seed, y = proportion)) +
    geom_point() +
    geom_line() +
    xlab("Tournament seed (smaller numbers indicate higher rank)")+
    ylab("Proportion reaching the top eight") +
    ggtitle("Top-Eight Proportion by Tournament Seed") +
    theme(text = element_text(size = 14))
seed_plot
```

Visualization 2: This plot compares regular-season win-percentage distributions between teams that reached the top eight and those that did not. Separation between the curves would suggest that win percentage may help predict the outcome, while overlap would indicate imperfect separation. Each curve is normalized separately, so the plot compares distribution shapes rather than group sizes

```r
# Compare win-percentage distributions between the two outcome groups
win_percentage_plot <- tidy_march_madness |>
    ggplot(aes(x = reg_wins_pct, colour = top_eight)) +
    geom_density() +
    xlab("Regular-season win percentage")+
    ylab("Density") +
    labs(colour = "Reached top eight \nFALSE = No; TRUE = Yes") +
    ggtitle("Regular-Season Win Percentage by Tournament Outcome") +
    theme(text = element_text(size = 14))

win_percentage_plot
    
```
